```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        
        if y_hat == 1 and y[j] == 0:
            weights = weights - lr*X[j]
        elif y_hat == 0 and y[j] == 1:
            weights = weights + lr*X[j]
            
    return weights[1:], weights[0]

def step(z):
    return 1 if z>0 else 0
```

```python
coef_, intercept_ = perceptron(X,y)

print(coef_)
print(intercept_)

m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
xx_input = np.linspace(-3,3,100)
yy_input = m*xx_input + b

plt.figure(figsize=(10,6))
plt.plot(xx_input,yy_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

```python
def perceptron(X, y):
    m = []
    b = []
    
    X = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(200):
        j = np.random.randint(0, 100)
        y_hat = step(np.dot(X[j], weights))
        weights = weights + lr * (y[j] - y_hat) * X[j]
        
        m.append(-(weights[1] / weights[2]))
        b.append(-(weights[0] / weights[2]))
        
    return m, b
```

```python
m, b = perceptron(X, y)
```

```python
%matplotlib inline
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
x_i = np.arange(-3, 3, 0.1)
y_i = x_i * m[0] + b[0]
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100)
line, = ax.plot(x_i, y_i, 'r-', linewidth=2)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
x_i = np.arange(-3, 3, 0.1)
y_i = x_i * m[0] + b[0]

ax.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100)
line, = ax.plot(x_i, y_i, 'r-', linewidth=2)

def update(i):
    y_i = x_i * m[i] + b[i]
    line.set_ydata(y_i)
    return line,

anim = FuncAnimation(fig, update, frames=len(m), interval=200, blit=True)

# Render using JSHTML instead of HTML5 video
HTML(anim.to_jshtml())
```
